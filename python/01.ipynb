```python
# 2. 变量：给数据起名字
name = '箐箐'      # str  字符串
age = 25          # int  整数
height = 1.63     # float 浮点数
is_student = True # bool 布尔值

print(name, age, height, is_student)
```

```python
# 3. 用 type() 查看每个变量的类型
print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

```python
# 4. 基本运算
a = 10
b = 3
print('加法', a + b)
print('除法', a / b)
print('整除', a // b)
print('取余', a % b)
print('乘方', a ** b)
```

```python
# 5. f-string 格式化输出
print(f'{name} 今年 {age} 岁，身高 {height} 米')
```

```python
# 6. 当堂练习：BMI 计算器
# BMI = 体重(kg) / 身高(m) 的平方
# 把下面的数字改成你自己的试试
weight = 55
height = 1.63
bmi = weight / height ** 2
print(f'BMI = {bmi:.1f}')
```
